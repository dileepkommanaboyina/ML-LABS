```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn import datasets
iris = datasets.load_iris()
print(iris)
```

```python
data = pd.DataFrame({
    "sepal_length": iris.data[:, 0],
    "sepal_width": iris.data[:, 1],
    "petal_length": iris.data[:, 2],
    "petal_width": iris.data[:, 3],
    "output": iris.target
})
```

```python
data
```

```python
#split dataset into train nad test
from sklearn.model_selection import train_test_split
```

```python
#Features and output splitting
X = data[['sepal_length','sepal_width','petal_length','petal_width']]
Y = data['output']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2, random_state=42)
```

```python
#model define
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators = 100)
```

```python
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
# classification score
from sklearn.metrics import accuracy_score
accuracy_score(Y_test, y_pred)
```

```python
y_pred = model.predict(X_train)
from sklearn.metrics import accuracy_score
accuracy_score(Y_train, y_pred)
```

```python
# Plot the iris data by class
plt.figure(figsize=(8, 6))
for species in sorted(data['output'].unique()):
    subset = data[data['output'] == species]
    plt.scatter(
        subset['petal_length'],
        subset['petal_width'],
        label=iris.target_names[species],
        alpha=0.8
    )

plt.title('Iris Dataset: Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.legend()
plt.grid(True)
plt.show()

# Model accuracy trend for different number of trees
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

n_estimators = list(range(50, 501, 50))
train_acc = []
test_acc = []

for n in n_estimators:
    model = RandomForestClassifier(n_estimators=n, random_state=42)
    model.fit(X_train, Y_train)
    train_acc.append(accuracy_score(Y_train, model.predict(X_train)))
    test_acc.append(accuracy_score(Y_test, model.predict(X_test)))

plt.figure(figsize=(8, 5))
plt.plot(n_estimators, train_acc, marker='o', label='Train Accuracy', linewidth=2)
plt.plot(n_estimators, test_acc, marker='s', label='Test Accuracy', linewidth=2)
plt.title('Random Forest Accuracy vs Number of Trees')
plt.xlabel('Number of Trees (n_estimators)')
plt.ylabel('Accuracy')
plt.xticks(n_estimators)
plt.ylim(0, 1.1)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
